```python
import pandas as pd
from sklearn import preprocessing
from sklearn.cluster import KMeans
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tests = [(pd.read_csv(filename)) for filename in ["TestSegments\Baseline1.csv", "TestSegments\Baseline2.csv", "TestSegments\Bump1.csv", "TestSegments\DoubleBump1.csv", "TestSegments\SmallLargeBump1.csv","TestSegments\SmallLargeBump2.csv"]]
```

```python
for i in range(len(tests)):
    tests[i]['time'] = pd.to_datetime(tests[i]['time'])
```

```python
def kmeans_fit(df, n):
    df_norm = preprocessing.normalize(df[['acc_z', 'gyro_y', 'pitch', 'roll']])
    model = KMeans(n_clusters=n, random_state=42, n_init=10)
    model.fit(df_norm)
    df['labels'] = model.labels_
    plot = sns.scatterplot(data=df, x='time', y='acc_z', hue='labels')
    plot.set_xticklabels(plot.get_xticklabels(), rotation=45)
    
    sns.pairplot(data=df, x_vars=['acc_z', 'gyro_y', 'pitch', 'roll'], y_vars=['acc_z', 'gyro_y', 'pitch', 'roll'], hue='labels')
    
```

# Baseline Tests

```python
kmeans_fit(tests[0], 4)
```

```python
kmeans_fit(tests[1], 4)
```
